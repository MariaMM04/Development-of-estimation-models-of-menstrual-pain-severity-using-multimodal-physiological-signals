```python
import pandas as pd
import numpy as np
import seaborn as sns  
import matplotlib.pyplot as plt

columns_to_study = ['id', 'study_interval', 'bpm']
df = pd.read_csv(r"C:\Users\Maria\TFG\Missing values\hr_finalvalues.csv", usecols=columns_to_study)
df_heartrate= df[~df['study_interval'].isin([2024])]

plt.figure(figsize=(30, 15))
sns.boxplot(data=df_heartrate, x='id', y='bpm')
    
plt.title('Heart Rate Distribution by Participant ID', fontsize=30, pad=20, weight='bold')
plt.xlabel('Participant ID', fontsize=26, labelpad=15)
plt.ylabel('Heart Rate Value (bpm)', fontsize=26, labelpad=15)
plt.xticks(rotation=45, fontsize=20) 
plt.yticks(fontsize=20)
plt.show()
```

```python
columns_to_study = ['id', 'study_interval', 'day_in_study', 'bpm']
df = pd.read_csv(r"C:\Users\Maria\TFG\Missing values\hr_finalvalues.csv", usecols=columns_to_study)
df_filtered = df[~df['study_interval'].isin([2024])]

columns_to_study_hormones = ['id', 'study_interval', 'day_in_study', 'phase']
df_phases = pd.read_csv(r"C:\Maria\TFG\Levels of pain_Hormones\hormones_and_selfreport.csv", usecols=columns_to_study_hormones)
df_filtered_hormones = df_phases[~df_phases['study_interval'].isin([2024])]


df_final = pd.merge(df_filtered, df_filtered_hormones,on=['id', 'study_interval','day_in_study'])

plt.figure(figsize=(10, 6))
sns.boxplot(data=df_final, x='phase', y='bpm')

plt.title('Heart Rate Across Menstrual Phases')
plt.xlabel('Menstrual Cycle Phase')
plt.ylabel('Heart Rate (Bpm)')
plt.xticks(rotation=45)

plt.show()
```
