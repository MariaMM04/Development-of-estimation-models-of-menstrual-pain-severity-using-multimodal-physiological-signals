```python
import pandas as pd
import numpy as np
import seaborn as sns  
import matplotlib.pyplot as plt
columns_to_study = ['id','study_interval','day_in_study','bpm']
df_bpm = pd.read_csv(r"C:\Users\Maria\TFG\Missing values\hr_finalvalues.csv", usecols=columns_to_study)
df_hr= df_bpm[~df_bpm['study_interval'].isin([2024])]

def graph_hr_per_day (file_temp, id):

    df_id= file_temp[file_temp['id']==id].copy()
    plt.figure(figsize=(30, 15))
    sns.boxplot(data=df_id, x='day_in_study', y='bpm')
        
    plt.title('Heart Rate values by Participant ID', fontsize=30, pad=20, weight='bold')
    plt.xlabel('Participant ID', fontsize=26, labelpad=15)
    plt.ylabel('Heart Rate values (bpm)', fontsize=26, labelpad=15)
    plt.xticks(rotation=45, fontsize=15) 
    plt.yticks(fontsize=20) 
    plt.show()
```
