```python
import pandas as pd
import numpy as np
import seaborn as sns  
import matplotlib.pyplot as plt

columns_to_study_lh = ['id', 'study_interval', 'lh', 'estrogen']
df_h = pd.read_csv(r"C:\Users\Maria\TFG\Missing values_per day\hormones_nomissing.csv", usecols=columns_to_study_lh)
df_hormones= df_h[~df_h['study_interval'].isin([2024])]

plt.figure(figsize=(30, 15))
sns.boxplot(data=df_hormones, x='id', y='lh')
    
plt.title('LH Distribution by Participant ID', fontsize=30, pad=20, weight='bold')
plt.xlabel('Participant ID', fontsize=26, labelpad=15)
plt.ylabel('LH Value (mIU/mL)', fontsize=26, labelpad=15)
plt.xticks(rotation=45, fontsize=20) 
plt.yticks(fontsize=20)
plt.show()

plt.figure(figsize=(30, 15))
sns.boxplot(data=df_hormones, x='id', y='estrogen')
    
plt.title('Estrogen Distribution by Participant ID', fontsize=30, pad=20, weight='bold')
plt.xlabel('Participant ID', fontsize=26, labelpad=15)
plt.ylabel('Estrogen Value (ng/mL)', fontsize=26, labelpad=15)
plt.xticks(rotation=45, fontsize=20) 
plt.yticks(fontsize=20) 
plt.show()
```
