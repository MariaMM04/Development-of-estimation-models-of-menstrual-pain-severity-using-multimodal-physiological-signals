```python
import pandas as pd
import numpy as np
import seaborn as sns  
import matplotlib.pyplot as plt
columns_to_study_diff = ['id', 'study_interval', 'day_in_study','temperature_diff_from_baseline']
df_diff = pd.read_csv(r"C:\Maria\TFG\Temperature\wrist_temperature.csv", usecols=columns_to_study_diff)
df_temperature_diff= df_diff[~df_diff['study_interval'].isin([2024])]

def graph_diff_from_baseline (file_temp, id):

    df_id= file_temp[file_temp['id']==id].copy()
    plt.figure(figsize=(30, 15))
    sns.boxplot(data=df_id, x='day_in_study', y='temperature_diff_from_baseline')
    plt.title('Temperature baseline deviation by Participant ID', fontsize=30, pad=20, weight='bold')
    plt.xlabel('Participant ID', fontsize=26, labelpad=15)
    plt.ylabel('Temperature baseline deviation (°C)', fontsize=26, labelpad=15)
    plt.xticks(rotation=45, fontsize=20) 
    plt.yticks(fontsize=20)
    plt.show()
```
