```python
import pandas as pd
import numpy as np
import seaborn as sns  
import matplotlib.pyplot as plt

columns_to_study = ['id', 'study_interval', 'nightly_temperature']
df = pd.read_csv(r"C:\Maria\TFG\Temperature\computed_temperature.csv", usecols=columns_to_study)
df_temperature= df[~df['study_interval'].isin([2024])]

plt.figure(figsize=(30, 15))
sns.boxplot(data=df_temperature, x='id', y='nightly_temperature')
    
plt.title('Temperature Distribution by Participant ID', fontsize=30, pad=20, weight='bold')
plt.xlabel('Participant ID', fontsize=26, labelpad=15)
plt.ylabel('Temperature Value (Cº)', fontsize=26, labelpad=15)
plt.xticks(rotation=45, fontsize=20) 
plt.yticks(fontsize=20) 
plt.show()
```

```python
df = pd.read_csv(r"C:\Maria\TFG\Temperature\computed_temperature.csv")
df_filtered = df[~df['study_interval'].isin([2024])]
df_phases = pd.read_csv(r"C:\Maria\TFG\Levels of pain_Hormones\hormones_and_selfreport.csv")

df_final = pd.merge(df_filtered, df_phases,left_on=['id', 'sleep_end_day_in_study'], right_on=['id', 'day_in_study'])

plt.figure(figsize=(10, 6))
sns.boxplot(data=df_final, x='phase', y='nightly_temperature')

plt.title('Temperature Across Menstrual Phases')
plt.xlabel('Menstrual Cycle Phase')
plt.ylabel('Temperature (ºC)')
plt.xticks(rotation=45)

plt.show()
```
