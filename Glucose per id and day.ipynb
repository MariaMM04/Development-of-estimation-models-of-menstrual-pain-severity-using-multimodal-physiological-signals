```python
import pandas as pd
import numpy as np
import seaborn as sns  
import matplotlib.pyplot as plt
columns_to_study = ['id', 'day_in_study','glucose_value']
df_glucose = pd.read_csv(r"C:\Users\Maria\TFG\Missing values\glucose_final_cleaned.csv", usecols=columns_to_study)

def graph_glucose_per_day (file_temp, id):

    df_id= file_temp[file_temp['id']==id].copy()
    plt.figure(figsize=(30, 15))
    sns.boxplot(data=df_id, x='day_in_study', y='glucose_value')
        
    plt.title('Glucose values by Participant ID', fontsize=30, pad=20, weight='bold')
    plt.xlabel('Participant ID', fontsize=26, labelpad=15)
    plt.ylabel('Glucose values (mmol/L)', fontsize=26, labelpad=15)
    plt.xticks(rotation=45, fontsize=20) 
    plt.yticks(fontsize=20) 
    plt.show()
```
